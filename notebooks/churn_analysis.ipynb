```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt 
import seaborn as sns
```

```python
data = pd.read_csv("../Data/churn.csv")
```

```python
data["TotalCharges"] = pd.to_numeric(data["TotalCharges"], errors="coerce")
data["TotalCharges"] = data["TotalCharges"].fillna(0)
```

```python
data = data.drop("customerID",axis = 1)
```

```python
data.head()
```

```python
X = data.drop("Churn",axis = 1)
Y = data["Churn"]
```

# handle categorical value 

```python
from sklearn.preprocessing import LabelEncoder ,OneHotEncoder 
```

```python
label_encoder = LabelEncoder()
OneHot_Encoder = OneHotEncoder()
```

```python
label_columns = ["gender","Partner","Dependents","PhoneService","OnlineSecurity","OnlineBackup","DeviceProtection","TechSupport","StreamingTV","StreamingMovies","PaperlessBilling"]
onehot_columns = ["MultipleLines","InternetService","Contract","PaymentMethod"]
```

```python
for col in X[label_columns]:
    X[col] = label_encoder.fit_transform(X[col])
```

```python
for col in onehot_columns:
    encoder = OneHotEncoder(sparse_output=False)
    onehot_encoded = encoder.fit_transform(X[[col]])
    encoded_df = pd.DataFrame(
        onehot_encoded,
        columns=[f"{col}_{cat}" for cat in encoder.categories_[0]],
        index=X.index
    )
    X = pd.concat([X, encoded_df], axis=1)
    if col in X.columns:
        X.drop(columns=[col], inplace=True)
```

# split the data into training and test data 

```python
from sklearn.model_selection import train_test_split 
```

```python
X_train, X_test, y_train, y_test = train_test_split (
    X,
    Y,
    random_state = 42,
    test_size = 0.2
)
```

# handle numerical columns 

```python
from sklearn.preprocessing import StandardScaler 
```

```python
scaler = StandardScaler()
```

```python
numeric_cols = ["tenure", "MonthlyCharges", "TotalCharges"]

X_train_scaled = X_train.copy()
X_test_scaled = X_test.copy()

X_train_scaled[numeric_cols] = scaler.fit_transform(X_train[numeric_cols])
X_test_scaled[numeric_cols] = scaler.transform(X_test[numeric_cols])
```

# model training 

```python
from sklearn.linear_model import LogisticRegression 
from sklearn.tree import DecisionTreeClassifier 
from sklearn.ensemble import RandomForestClassifier 
```

```python
linear_model = LogisticRegression()
tree_model = DecisionTreeClassifier()
forest_model = RandomForestClassifier()
```

```python
TREE_MODEL = tree_model.fit(X_train_scaled, y_train)
FOREST_MODEL = forest_model.fit(X_train_scaled, y_train)
LINEAR_MODEL = linear_model.fit(X_train_scaled, y_train)
```

```python
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
```

```python
tree_accuracy = accuracy_score(y_test, TREE_MODEL.predict(X_test_scaled))
forest_accuracy = accuracy_score(y_test, FOREST_MODEL.predict(X_test_scaled))
linear_accuracy = accuracy_score(y_test, LINEAR_MODEL.predict(X_test_scaled))
```

```python
print("Decision Tree Accuracy:", tree_accuracy)
print("Random Forest Accuracy:", forest_accuracy)
print("Logistic Regression Accuracy:", linear_accuracy)
```
